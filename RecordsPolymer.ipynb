```python
from polymer.load_tfrecords_croppy import (
    compute_global_stats, 
    verify_crops_reconstruction, 
    get_jax_iterator
)

# 1. Verify cropping works (simple, no standardization)
results = verify_crops_reconstruction(
    "/data/tfrecords_pe/*.tfrecord",
    crop_size=256,
    stride=192
)

# 2. Compute global stats
global_mean, global_std = compute_global_stats("/data/tfrecords_pe/*.tfrecord")

# 3. Training iterator with augmentations
train_iter = get_jax_iterator(
    "/data/tfrecords_pe/*.tfrecord",
    crop_size=256,
    stride=192,
    batch_size=32,
    shuffle=True,
    seed=42,
    augment=True,
    global_mean=global_mean,
    global_std=global_std,
    noise_std=0.02
)
batch = next(train_iter)
print(f"Batch shape: {batch['image'].shape}")

# # 4. Validation iterator (no augmentation)
# val_iter = get_jax_iterator(
#     "/data/tfrecords_pe_val/*.tfrecord",
#     crop_size=256,
#     stride=192,
#     batch_size=32,
#     shuffle=False,
#     augment=False,
#     global_mean=global_mean,
#     global_std=global_std
# )
```
